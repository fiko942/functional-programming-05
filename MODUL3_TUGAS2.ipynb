Pengolahan data pengiriman untuk perusahaan ekspedisi dengan pendekatan fungsional.

```python
from functools import reduce
from pprint import pprint

pengiriman = [
    {"resi": "A001", "berat": 2.5, "ongkir_perkg": 10000, "status": "Terkirim"},
    {"resi": "A002", "berat": 5, "ongkir_perkg": 12000, "status": "Dalam Proses"},
    {"resi": "A003", "berat": 1.2, "ongkir_perkg": 10000, "status": "Terkirim"},
    {"resi": "A004", "berat": 10, "ongkir_perkg": 9000, "status": "Terkirim"},
    {"resi": "A005", "berat": 0.5, "ongkir_perkg": 15000, "status": "Batal"},
    {"resi": "A006", "berat": 3.8, "ongkir_perkg": 11000, "status": "Dalam Proses"},
    {"resi": "A007", "berat": 7.2, "ongkir_perkg": 9500, "status": "Terkirim"},
    {"resi": "A008", "berat": 4.5, "ongkir_perkg": 10500, "status": "Terkirim"},
    {"resi": "A009", "berat": 6.0, "ongkir_perkg": 9800, "status": "Batal"},
    {"resi": "A010", "berat": 9.1, "ongkir_perkg": 8700, "status": "Terkirim"},
    {"resi": "A011", "berat": 2.0, "ongkir_perkg": 12500, "status": "Dalam Proses"},
    {"resi": "A012", "berat": 1.5, "ongkir_perkg": 14000, "status": "Terkirim"},
    {"resi": "A013", "berat": 8.3, "ongkir_perkg": 9200, "status": "Terkirim"},
    {"resi": "A014", "berat": 0.8, "ongkir_perkg": 16000, "status": "Batal"},
    {"resi": "A015", "berat": 5.6, "ongkir_perkg": 9900, "status": "Terkirim"},
]

def status_terkirim(entry):
    return entry["status"] == "Terkirim"


def filter_pengiriman_terkirim(data):
    # filter: menyaring data yang berstatus terkirim
    return list(filter(status_terkirim, data))


def hitung_total(entry):
    total = entry["berat"] * entry["ongkir_perkg"]
    return {**entry, "total_ongkir": total}


def tambah_total_ongkir(data):
    # map: menambahkan field total ongkir ke setiap entri baru
    print("Hitung total", hitung_total(data[0]))
    return list(map(hitung_total, data))


def akumulasi_ongkir(acc, entry):
    return acc + entry["total_ongkir"]


def hitung_total_pendapatan(data):
    # reduce: melipat daftar total ongkir menjadi satu nilai pendapatan
    return reduce(akumulasi_ongkir, data, 0)


def pengiriman_tanpa_batal(data):
    # list comprehension: membentuk koleksi baru tanpa status Batal
    return [entry for entry in data if entry["status"] != "Batal"]


def total_berat_rekursif(data):
    # rekursi: menjumlahkan berat pengiriman secara deklaratif tanpa loop
    if not data:
        return 0
    pertama, *sisanya = data
    return pertama["berat"] + total_berat_rekursif(sisanya)


pengiriman_terkirim = filter_pengiriman_terkirim(pengiriman)
pengiriman_dengan_total = tambah_total_ongkir(pengiriman_terkirim)
total_pendapatan = hitung_total_pendapatan(pengiriman_dengan_total)
pengiriman_non_batal = pengiriman_tanpa_batal(pengiriman)
total_berat_non_batal = total_berat_rekursif(pengiriman_non_batal)

print("=== Skenario 1: Pengiriman Terkirim ===")
pprint(pengiriman_terkirim)

print("\n=== Skenario 2: Tambah Total Ongkir ===")
pprint(pengiriman_dengan_total)

print("\n=== Skenario 3: Total Pendapatan ===")
print(total_pendapatan)

print("\n=== Skenario 4: Data Non-Batal & Total Berat ===")
pprint(pengiriman_non_batal)
print("Total berat semua pengiriman non-batal:", total_berat_non_batal)

assert all("total_ongkir" not in entry for entry in pengiriman)
pprint(pengiriman)
```
